## Learning objectives

- Design narrow tool schemas.
- Validate inputs before execution.
- Keep application control in a deterministic workflow.

```python
def get_price_snapshot(ticker: str) -> dict[str, str | float]:
    """Return a deterministic teaching fixture before using live market data."""
    fixtures = {"ACME": {"ticker": "ACME", "price": 101.25, "currency": "USD"}}
    if ticker not in fixtures:
        raise ValueError(f"Unknown teaching ticker: {ticker}")
    return fixtures[ticker]

get_price_snapshot("ACME")
```
